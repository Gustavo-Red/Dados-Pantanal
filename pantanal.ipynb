```python
import pandas

# comando para ler um arquivo:  pandas.read_[tipo do arquivo]("[nome do arquivo]")
# neste caso o comando é: pandas.read_csv("dados_pantanal.csv")

tabela = pandas.read_csv("dados_pantanal.csv")

# comando para mostrar tabelas de maneira "bonitinha": display([base de dados, em questão, que foi lida com pandas.read])
# neste caso o comando é: display(tabela)

display(tabela)
```

```python
display(tabela.info())
```

```python
# Para preencher os valores vazios estimando o valor com base nos vizinhos
# Use o comando [Base de dados lida anteriormente].interpolate()
# Nesse caso seria: tabela = tabela.interpolate()
# Porém, para o código rodar perfeitamente, é necessário um ajuste. 
# Pois o Pandas não sabe calcular a "média" entre textos, então é necessário ignorar a coluna de data na hora de interpolar ou focar apenas nas colunas numéricas.

# 1. Transformamos a coluna data em um formato que o Pandas entenda como tempo
# A coluna "data" ta em formato de string, o que resulta em um erro
# Use o comando pandas.to_datetime([base de dados lida anteriormente]["[coluna de datas da tabela]"])


tabela["data"] = pandas.to_datetime(tabela["data"])

#apos isso, a coluna "data" será do Dtype -exibido no display(tabela.info())- "datetime64". Como é um formato de tempo, o Pandas irá ignorar essa coluna durante o cálculo matemático, evitando o erro

# 2. Fazemos a interpolação apenas nas colunas que são números
tabela = tabela.interpolate()

# 3. Exibimos a tabela pronta
display(tabela)
display(tabela.info())
```

```python
#################       Renomeando as colunas      #####################

#Renomearemos as colunas para aparecerem os "nomes bonitos" nos gráficos


# Renomeando via código:
tabela = tabela.rename(columns={
    "data": "Data",
    "temperatura_c": "Temperatura (°C)",
    "nivel_rio_m": "Nível do rio (m)",
    "ndvi": "Índice de Vegetação (NDVI)"
})


display(tabela)
```

```python
##################      Esboçando em gráficos      #####################

#Obs: Para isso será utilizada a biblioteca plotly

import plotly.express as px  # <-- Sempre importe a biblioteca plotly assim

#comando para criar gráficos: 
# px.[tipo de gráfico que você quer criar]([Base de dados lida anteriormente], x="coluna que voce quer que seja o eixo x", y="coluna que voce quer que seja o eixo y")

#comando que exibe o gráfico: [Gráfico feito anteriormente].show()


grafico = px.line(tabela, x="Data", y="Temperatura (°C)")

display(grafico) # <- Grafico que será exibido só para mostrar como o comando display() funciona
```

```python
### Criando os gráficos possíveis em função do tempo ###

for coluna in tabela.columns:
    if coluna == "Data":
        continue
    #comando para fazer a média dos dados de uma coluna (biblioteca: pandas)
    # [base de dados lida anteriormente][coluna].mean()
    # Neste caso: tabela[coluna].mean()

    media = tabela[coluna].mean()


    grafico = px.line(tabela, x="Data", y=coluna, title=f"{coluna} ao decorrer dos dias (Média: {media:.2f})", markers=True)
    
    grafico.update_traces(line=dict(width=3)) # <- Muda a espessura da linha
    grafico.update_layout(template="plotly_dark") # <- Muda a estética do fundo
    grafico.update_layout(xaxis=dict(dtick="D1")) # <- Força a mostrar cada dia
    grafico.show()
```
